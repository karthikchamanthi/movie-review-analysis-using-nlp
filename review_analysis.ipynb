```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = df.drop_duplicates()
```

```python
df['sentiment'].value_counts()
```

```python
df['review'].value_counts()
```

```python
df['review'] = df['review'].apply(lambda x:x.lower())
```

```python
df['review'].head()
```

```python
import string

def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['review'] = df['review'].apply(remove_punc)
```

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['review'] = df['review'].apply(remove_numbers)
```

```python
def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new

df['review'] = df['review'].apply(remove_emojis)
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
stop_words
```

```python
def remove(txt):
  words = txt.split()
  cleaned = []
  for i in words:
    if not i in stop_words:
      cleaned.append(i)

  return ' '.join(cleaned)
```

```python
df['review'] = df['review'].apply(remove)
```

```python
df['sentiment'] = df['sentiment'].map({
    'negative': 0,
    'positive': 1
})

df['sentiment'].value_counts()
```

```python
df['sentiment'].head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['review'], df['sentiment'], test_size=0.20, random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
```

```python
bow = CountVectorizer()
```

```python
Xtrain = bow.fit_transform(X_train)
```

```python
Xtest = bow.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
lg = LogisticRegression()
```

```python
lg.fit(Xtrain,y_train)
```

```python
lg.score(Xtest,y_test)
```

```python
y_pred = lg.predict(Xtest)

print(y_pred[:10])
```

```python
from sklearn.metrics import accuracy_score, f1_score, classification_report
```

```python
accuracy = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Accuracy:", round(accuracy, 4))
print("F1-score:", round(f1, 4))
```

```python
print(classification_report(
    y_test,
    y_pred,
    target_names=['Negative', 'Positive']
))
```
